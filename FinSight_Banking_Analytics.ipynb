```python
df = df.drop_duplicates()
```

```python
df.replace("unknown", np.nan, inplace=True)
```

```python
df.isnull().sum()
```

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[18, 25, 35, 45, 55, 65, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65", "65+"]
)
```

```python
df["balance_segment"] = pd.cut(
    df["balance"],
    bins=[-10000, 0, 1000, 5000, 10000, 100000],
    labels=["Negative", "Low", "Medium", "High", "Very High"]
)
```

```python
df["loan_status"] = df["loan"] + "_" + df["housing"]
```

```python
df["deposit"].value_counts()
df["deposit"].value_counts(normalize=True) * 100
df["balance"].mean()
df["age"].mean()
```

```python
df.to_sql("bank_customers", conn, if_exists="replace", index=False)
```

```python
pd.read_sql("SELECT * FROM bank_customers LIMIT 5", conn)
```

```python
query = "SELECT COUNT(*) AS total_customers FROM bank_customers"
pd.read_sql(query, conn)
```

```python
query = """
SELECT 
ROUND(
    100.0 * SUM(CASE WHEN deposit='yes' THEN 1 ELSE 0 END) / COUNT(*),
    2
) AS conversion_rate
FROM bank_customers
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT ROUND(AVG(balance), 2) AS avg_balance
FROM bank_customers
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT ROUND(AVG(age), 2) AS avg_age
FROM bank_customers
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT
job,
COUNT(*) AS total_customers,
SUM(CASE WHEN deposit='yes' THEN 1 ELSE 0 END) AS conversions,
ROUND(
100.0 * SUM(CASE WHEN deposit='yes' THEN 1 ELSE 0 END)/COUNT(*),2
) AS conversion_rate
FROM bank_customers
GROUP BY job
ORDER BY conversion_rate DESC
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT
marital,
COUNT(*) AS customers,
SUM(CASE WHEN deposit='yes' THEN 1 ELSE 0 END) AS conversions
FROM bank_customers
GROUP BY marital
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT
education,
COUNT(*) AS customers,
SUM(CASE WHEN deposit='yes' THEN 1 ELSE 0 END) AS conversions
FROM bank_customers
GROUP BY education
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT
contact,
COUNT(*) AS customers,
SUM(CASE WHEN deposit='yes' THEN 1 ELSE 0 END) AS conversions
FROM bank_customers
GROUP BY contact
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT
month,
COUNT(*) AS customers,
SUM(CASE WHEN deposit='yes' THEN 1 ELSE 0 END) AS conversions
FROM bank_customers
GROUP BY month
"""
pd.read_sql(query, conn)
```

```python
query = """
SELECT
[default],
loan,
housing,
COUNT(*) AS customers
FROM bank_customers
GROUP BY [default], loan, housing
ORDER BY customers DESC
"""
pd.read_sql(query, conn)
```
